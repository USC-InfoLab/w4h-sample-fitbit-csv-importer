# Import Fitbit CSV sample package

Uses **w4h-sample-fitbit-csv** + your personal API key from the environment.

1. `cp .env.example .env` and set `W4H_API_KEY`
2. Ensure `W4H_SAMPLE_FITBIT_CSV_PATH` points at the sample repo
3. Set `W4H_DATASET_ID` to your catalog dataset id

```python
from w4h_sample_fitbit_csv_importer.client import W4HClient
from w4h_sample_fitbit_csv_importer.config import dataset_id, sample_package_path
from w4h_sample_fitbit_csv_importer.csv_import import import_all
from w4h_sample_fitbit_csv_importer.manifest import load_manifest

package_root = sample_package_path()
manifest = load_manifest(package_root)
client = W4HClient()
ds_id = dataset_id()
print("package:", package_root)
print("dataset:", ds_id)
```

```python
results = import_all(client, ds_id, manifest, package_root, mode="replace")
results
```
